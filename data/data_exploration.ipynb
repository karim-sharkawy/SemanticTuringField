```python
import pandas as pd

# Full file, not available on GitHub. Use download_data.py for link to full file
df = pd.read_csv(
    "C:/Users/karim/Downloads/STF/data/wiki_giga_2024_50_MFT20_vectors_seed_123_alpha_0.75_eta_0.075_combined.txt",
    sep=r"\s+",
    header=None
    )

#df
```

```python
words = df.iloc[:, 0].values
vectors = df.iloc[:, 1:].values.astype(float)

print(f"Vocabulary size: {len(words):,}\n")
print(f"Vector dimensions: {vectors.shape[1]}\n")
print(f"Sample: {df[0].sample(50).tolist()}")
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

word_to_idx = {w: i for i, w in enumerate(words)}

def nearest_words(word, k=5):
    idx = word_to_idx[word]
    vec = vectors[idx].reshape(1, -1)

    sims = cosine_similarity(vec, vectors)[0]

    top = np.argsort(sims)[::-1][1:k+1]

    print(f"Words for {word}:")
    #return [(words[i], sims[i]) for i in top] # if you also want the similarity
    return [words[i] for i in top]

random_words = df[0].sample(7).tolist()
for word in random_words:
    print(nearest_words(word))
    print()
```

```python
import random

def analogy(a, b, c, k=10):
    target = (
        vectors[word_to_idx[b]]
        - vectors[word_to_idx[a]]
        + vectors[word_to_idx[c]]
    )

    sims = cosine_similarity(
        target.reshape(1, -1),
        vectors
    )[0]

    top = np.argsort(sims)[::-1][:k]

    #return [(words[i], sims[i]) for i in top] # if you also want the similarity
    return [words[i] for i in top]

for _ in range(5):
    word1, word2, word3 = random.sample(list(words), 3) # can also use random_words

    print(f"{word2} - {word1} + {word3}")
    print(analogy(word1, word2, word3))
    print()
```
